```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
corpus = ["Natural language processing (NLP) is a subfield of artificial intelligence (AI) that focuses on the interaction between computers and humans through natural language.",
          "NLP enables computers to understand, interpret, and generate human language in a way that is both meaningful and useful.", 

          "Machine learning algorithms are often used in NLP to analyze and model language data, allowing for tasks such as sentiment analysis, language translation, and text summarization.",
          "Applications of NLP can be found in various industries, including healthcare, finance, customer service, and more. It plays a crucial role in improving communication between humans and machines.",
          "NLP techniques involve tasks such as tokenization, part-of-speech tagging, named entity recognition, and syntactic parsing, which help in extracting meaningful information from text data."]
```

```python
print("Corpus")
print("-"*50)

for i, doc in enumerate(corpus,1):
    print(f"Document {i}: {doc}")
```

```python
vectorizer = TfidfVectorizer()
```

```python
vectorizer.fit(corpus)
```

```python
print(vectorizer.vocabulary_)
```

```python
features = vectorizer.get_feature_names_out()

print(features)
```

```python
tfidf_matrix = vectorizer.transform(corpus)
```

```python
matrix = tfidf_matrix.toarray()

print(matrix)
```

```python
df = pd.DataFrame(

matrix,

columns=features

)

print(df)
```

```python
print(df.iloc[0])
```

```python
print(df["machine"])
```

```python
df.to_csv("tfidf_matrix.csv",index=False)

print("TF-IDF Matrix Saved Successfully")
```
