# Practical 1: Text Preprocessing Pipeline Implementation

**Concept:** Raw text data is almost never clean. It contains punctuation, numbers, stopwords, inconsistent capitalization, and word variations. Before feeding text into any NLP model, we must pass it through a **preprocessing pipeline** that converts it into a clean, standardized form.

This notebook builds a complete preprocessing pipeline step-by-step:
1. Sentence & Word Tokenization
2. Lowercasing
3. Removing Punctuation & Numbers
4. Removing Stopwords
5. Stemming
6. Lemmatization
7. Saving the Final Clean Corpus

## Import Required Libraries
We use **NLTK** (Natural Language Toolkit) for tokenization, stopwords, stemming and lemmatization.

```python
# Tokenization
from nltk.tokenize import sent_tokenize, word_tokenize

# Stopwords
from nltk.corpus import stopwords

# Stemming
from nltk.stem import PorterStemmer

# Lemmatization
from nltk.stem import WordNetLemmatizer

# NLTK
import nltk

# String Operations
import string

# Regular Expressions
import re
```

## Download Required NLTK Resources
These small data packages are required for tokenization, stopword lists and WordNet.

```python
# Download required NLTK data
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

## Create the Corpus
We create a small sample text (a corpus) that contains uppercase words, punctuation, numbers and stopwords — everything a real-world preprocessing pipeline has to handle.

```python
text = """
Machine Learning is one of the most important fields of Artificial Intelligence.
It enables computers to learn from data without being explicitly programmed.
Today, Machine Learning is used in healthcare, banking, education, e-commerce,
cybersecurity, self-driving cars, and recommendation systems.

Over 2000 students study Machine Learning because it helps solve real-world problems.
However, before applying any machine learning algorithm,
the textual data must be cleaned and preprocessed.
Text preprocessing improves the quality of data by removing unnecessary
symbols, stopwords, punctuation, numbers, and extra spaces.
"""
```

## Display Original Text

```python
print("Original Corpus")
print("-" * 50)
print(text)
```

## Sentence Tokenization
**Sentence Tokenization** splits a large block of text into individual sentences, using punctuation as boundaries.

```python
# Split text into sentences
sentences = sent_tokenize(text)

print("Sentence Tokenization")
print("-" * 50)

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")
```

## Word Tokenization
**Word Tokenization** breaks each sentence into individual words (tokens). Punctuation marks are kept as separate tokens by NLTK.

```python
# Split text into word tokens
words = word_tokenize(text)

print("Word Tokenization")
print("-" * 50)
print(words)
```

## Convert into Lowercase
**Lowercasing** makes "Machine" and "machine" the same token. This reduces the vocabulary size and prevents the model from treating capitalized and lowercase versions as different words.

```python
# Convert every token to lowercase
lower_words = [word.lower() for word in words]

print("Lowercase Words")
print("-" * 50)
print(lower_words)
```

## Remove Punctuation
Punctuation symbols like `.`, `,` and `-` usually carry little meaning for many NLP tasks (unless they matter, e.g. in sentence boundaries). We filter them out using Python's `string.punctuation`.

```python
# Keep only tokens that are NOT punctuation symbols
words_no_punctuation = [
    word for word in lower_words
    if word not in string.punctuation
]

print("After Removing Punctuation")
print("-" * 50)
print(words_no_punctuation)
```

## Remove Numbers
Numeric values (like "2000") are rarely useful as features in many text tasks. `isalpha()` keeps only tokens made entirely of alphabetic characters.

```python
# Keep only alphabetic tokens (removes numbers and symbols)
words_no_numbers = [
    word for word in words_no_punctuation
    if word.isalpha()
]

print("After Removing Numbers")
print("-" * 50)
print(words_no_numbers)
```

## Remove Stopwords
**Stopwords** are very common words (like *is*, *the*, *for*) that add little meaning. Removing them focuses the data on the informative words.

```python
# Load the English stopword list
stop_words = set(stopwords.words("english"))

# Keep only words that are NOT stopwords
filtered_words = [
    word for word in words_no_numbers
    if word not in stop_words
]

print("After Stopword Removal")
print("-" * 50)
print(filtered_words)
```

## Stemming
**Stemming** reduces words to their base stem by chopping off affixes using crude heuristics. For example, *"learning"* becomes *"learn"*. Note: stems are not always real words (e.g. *"machin"*).

```python
# Porter Stemmer is the most common English stemmer
stemmer = PorterStemmer()

# Apply stemming to every word
stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("After Stemming")
print("-" * 50)
print(stemmed_words)
```

## Lemmatization
**Lemmatization** reduces words to their dictionary base form (lemma) using a vocabulary and morphological analysis. Unlike stemming, it always produces real words (e.g. *"studies"* → *"study"*).

```python
# WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Apply lemmatization to every word
lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("After Lemmatization")
print("-" * 50)
print(lemmatized_words)
```

## Create Final Corpus
All the cleaned tokens are joined back into a single continuous string — this is the **final clean corpus** ready to be used by NLP models.

```python
# Join all cleaned words into one string
final_corpus = " ".join(lemmatized_words)

print("Final Clean Corpus")
print("-" * 50)
print(final_corpus)
```

## Save Final Corpus
The cleaned corpus is written to a text file so it can be reused in later practicals (e.g. for Word2Vec training).

```python
# Save the clean corpus to a file
with open("final_corpus.txt", "w", encoding="utf-8") as file:
    file.write(final_corpus)

print("Corpus Saved Successfully!")
```
