# Practical 8: Rule-based Chatbot

**Concept:** A **chatbot** is a program that simulates a conversation with a human user. A **rule-based chatbot** answers without any machine learning: it matches the user's input against a set of hand-written rules (regex patterns, keywords or if-else conditions) and replies from a **predefined knowledge base**. If no rule matches, it falls back to a generic response.

In this practical we build a simple college-information chatbot that answers questions about courses, fees, timings and placement using regular expressions.


## Import Required Libraries
We only need the standard library `re` for regular-expression matching.


```python
import re
import string
```

## Text Normalization
Before matching, we clean the user's input: lowercase it, strip punctuation and collapse extra spaces. This makes the patterns match more reliably.


```python
def normalize(text):
    """Lowercase, remove punctuation, collapse spaces."""
    text = text.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    return re.sub(r"\s+", " ", text).strip()


# Quick test
print(normalize("  Hello,  How   Are You?? "))
print("-" * 50)
```

## Knowledge Base
The bot's **knowledge base** is a dictionary of topics and the facts it knows about each of them. A real chatbot would read this from a database; here it is hard-coded for a college.


```python
knowledge_base = {
    "about":    "I am CampusBot, a simple rule-based chatbot that answers questions about XYZ College.",
    "courses":  "We offer B.Tech in CSE, IT and ECE, plus an M.Tech in AI and a BBA program.",
    "fees":     "The annual tuition fee is ₹1,20,000 for B.Tech and ₹1,40,000 for M.Tech.",
    "admission": "Admissions open in May. Applications are accepted online until June 30.",
    "timings":  "The college is open from 9:00 AM to 5:00 PM, Monday to Friday.",
    "location": "XYZ College is located on University Road, near City Park, Pune.",
    "contact":  "You can reach us at info@xyzcollege.edu or call 020-1234-5678.",
    "faculty":  "Our faculty includes 45 PhD holders across all departments.",
    "placement": "Last year 92% of final-year students were placed, with an average package of ₹8.5 LPA.",
    "library":  "The central library has 50,000+ books and 24/7 reading halls.",
}
```

## Rule Engine (Regex Patterns)
Rules are stored as a list of `(pattern, response)` pairs. The patterns are **regular expressions** matched anywhere in the input (`re.search`), so the user can phrase a question in many ways. The first matching rule decides the answer.


```python
rules = [
    # (regex pattern, response text)
    (r"\b(hi|hello|hey|greetings)\b",            "Hello! How can I help you today?"),
    (r"what is your name|your name|who are you",   knowledge_base["about"]),
    (r"are you a (human|robot|bot|machine)",       "I am a bot! I run on simple if-else rules and regex, not on a human brain."),
    (r"how are you",                               "I am just a few lines of Python, so I am always feeling fine!"),
    (r"(courses|programs|subjects|syllabus)",      knowledge_base["courses"]),
    (r"(fees|fee|cost|tuition|price)",             knowledge_base["fees"]),
    (r"(admission|enroll|enrol|apply|application)", knowledge_base["admission"]),
    (r"(timing|timings|hours|open)",               knowledge_base["timings"]),
    (r"(location|address|campus|where)",           knowledge_base["location"]),
    (r"(contact|email|phone|call|reach)",          knowledge_base["contact"]),
    (r"(faculty|teachers|professor)",              knowledge_base["faculty"]),
    (r"(placement|job|career|internship|salary)",  knowledge_base["placement"]),
    (r"(library|books)",                           knowledge_base["library"]),
    (r"(thank|thanks)",                            "You're welcome! Anything else I can help with?"),
    (r"\b(bye|goodbye|see you|exit)\b",           "Goodbye! Have a great day."),
]
```

## The Response Engine
The `chatbot_response` function:
1. Normalizes the user input.
2. Tries every rule in order and returns the answer of the **first** match.
3. Falls back to a default answer if nothing matches.


```python
def chatbot_response(user_input):
    """Return the bot's reply for a given user input."""
    cleaned = normalize(user_input)

    if not cleaned:
        return "Please type a message so I can help you."

    for pattern, response in rules:
        if re.search(pattern, cleaned):
            return response

    # Fallback for unrecognized input
    return "Sorry, I did not understand that. Try asking about courses, fees, timings or placement."


# Quick unit-style test
print(chatbot_response("Can you tell me about the courses?"))
print(chatbot_response("What are the timings?"))
print(chatbot_response("Do you sell pizza?"))
print("-" * 50)
```

## Testing the Chatbot
We run a set of sample user messages and print the bot's reply for each one.


```python
sample_queries = [
    "Hello",
    "What is your name?",
    "What courses do you offer?",
    "How much is the fee?",
    "Where is the college located?",
    "What are the college timings?",
    "How can I contact the college?",
    "Tell me about placements",
    "Are there any books in the library?",
    "Thanks!",
    "Bye",
    "What is the meaning of life?",
]

print("Chatbot Demo")
print("-" * 50)
for query in sample_queries:
    print(f"User: {query}")
    print(f"Bot : {chatbot_response(query)}")
    print()
```

## Interactive Mode
When you run the cell below in Jupyter, it starts a live chat loop in the console. Type messages and press Enter; type `exit` to stop.


```python
# Set to True to start a live chat in the console
INTERACTIVE = False

if INTERACTIVE:
    print("CampusBot: Hello! Ask me about courses, fees, timings, placement... (type 'exit' to quit)")
    while True:
        user_input = input("You: ")
        if normalize(user_input) == "exit":
            print("CampusBot: Goodbye!")
            break
        print(f"CampusBot: {chatbot_response(user_input)}")
```

## Bonus: Keyword-Scoring Matcher
Regex rules fail as soon as the user rephrases a question. An alternative is to **score** each topic by how many of its keywords appear in the input, and answer with the best-scoring topic. This is still rule-based, but more tolerant to phrasing.


```python
keywords = {
    "courses":   ["course", "program", "subject", "study", "branch"],
    "fees":      ["fee", "cost", "tuition", "price", "pay"],
    "admission": ["admit", "enroll", "apply", "application", "seat"],
    "timings":   ["timing", "time", "hour", "open", "close"],
    "location":  ["where", "location", "address", "campus"],
    "contact":   ["contact", "email", "phone", "call", "number"],
    "placement": ["placement", "job", "career", "intern", "package"],
    "library":   ["library", "book", "read"],
}


def keyword_response(user_input):
    """Reply with the topic whose keywords match the input the most."""
    cleaned = normalize(user_input)
    best_topic, best_score = None, 0

    for topic, words in keywords.items():
        score = sum(1 for word in words if word in cleaned)
        if score > best_score:
            best_topic, best_score = topic, score

    if best_topic is None:
        return "Sorry, I did not understand that. Try asking about courses, fees, timings or placement."
    return knowledge_base[best_topic]


print("Keyword-Scoring Demo")
print("-" * 50)
for query in ["How much do I have to pay for tuition?",
              "Can I apply for admission next month?",
              "Where exactly are you located?",
              "Do you have library facilities?",
              "random gibberish here"]:
    print(f"User: {query}")
    print(f"Bot : {keyword_response(query)}")
    print()
```

## Summary

We built a **rule-based chatbot**:
1. Normalized user input (lowercase, strip punctuation).
2. Stored facts in a **knowledge base** dictionary.
3. Matched input against **regex rules** and answered from the knowledge base.
4. Added a keyword-scoring matcher as a more tolerant alternative.

This "matching + canned response" design is exactly how early chatbots (like ELIZA) worked, and it still powers simple FAQ bots today — no machine learning required.
