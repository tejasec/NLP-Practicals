# Original Corpus

```python
corpus = [
    "machine learning transform artificial intelligence",
    "machine learning use neural network",
    "deep learning subset machine learning"
]
```

# Tokenization & Vocabulary

```python
tokenized = [doc.split() for doc in corpus]

vocabulary = sorted(set(word for doc in tokenized for word in doc))

print(vocabulary)
```

# 1. Term Frequency (TF)
TF measures how frequently a term occurs in a document. 
Formula: `(Number of times term t appears in a document) / (Total number of words in that document)`

```python
tf_matrix = []

for doc in tokenized:
    tf_vector = []
    for word in vocabulary:
        # Term count in doc / Total words in doc
        tf = doc.count(word) / len(doc)
        tf_vector.append(tf)
    
    tf_matrix.append(tf_vector)

for row in tf_matrix:
    print(row)
```

# 2. Inverse Document Frequency (IDF)
IDF measures how important a term is. It weighs down frequent words (like "machine" or "learning") and scales up rare words.
Formula: `log(Total number of documents / Number of documents containing term t)`

```python
import math

N = len(corpus)
idf_vector = []

for word in vocabulary:
    # Count how many documents contain this word
    df = sum(1 for doc in tokenized if word in doc)
    
    # Standard IDF calculation
    idf = math.log10(N / df)
    idf_vector.append(round(idf, 4))

print(idf_vector)
```

# 3. Manual TF-IDF Matrix
Multiply the Term Frequency (TF) by the Inverse Document Frequency (IDF) for each word.

```python
tfidf_matrix = []

for tf_vector in tf_matrix:
    tfidf_vector = []
    for i in range(len(vocabulary)):
        # TF * IDF
        tfidf = tf_vector[i] * idf_vector[i]
        tfidf_vector.append(round(tfidf, 4))
    
    tfidf_matrix.append(tfidf_vector)

for row in tfidf_matrix:
    print(row)
```

# Scikit-Learn TfidfVectorizer
Note: Scikit-learn uses a slightly modified formula for TF-IDF (incorporating smoothing by adding 1 to the numerator and denominator of the IDF equation) and applies L2 normalization to the final vectors by default. Because of this, the final array values will look different than our simple manual calculation above, but the concept is exactly the same!

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
