## 1. Why Word2Vec? The Problem With Sparse Vectors

Machine-learning models cannot read text, so words must be turned into numbers. Practicals 1–3 used *sparse* representations:

- **One-Hot Encoding** — a vector of length $|V|$ (vocabulary size) with a single `1` and everything else `0`.
- **Bag of Words (BoW)** — how often each word occurs in a document, ignoring order.
- **TF-IDF** — BoW weighted by how *rare* a word is across documents.

| Limitation | What it means in practice |
| :--- | :--- |
| **Extreme dimensionality & sparsity** | A 50,000-word vocabulary makes every vector 50,000 numbers long, ~99.9% of them zero. Memory explodes and distances become meaningless — the *curse of dimensionality*. |
| **No semantic relationships** | In One-Hot / BoW space every distinct word is **orthogonal**, so `cosine("machine", "computer") = 0.0` even though they are near-synonyms. Words can only be *equal*, never *similar*. |
| **No context, no word order** | `"dog bites man"` and `"man bites dog"` produce identical BoW vectors. |

**Word2Vec (Mikolov et al., 2013)** fixes this by learning **dense, continuous, low-dimensional** vectors
(typically 50–300 numbers instead of 50,000 sparse ones) directly from co-occurrence statistics.

````python
# --- Purpose: prove numerically that One-Hot vectors carry no semantic information. ---
# Why it matters: this is the exact weakness Word2Vec repairs. If every pair of distinct words
# scores 0.0, no vector-based model can ever tell us that "machine" ~ "computer".

VOCAB = ["machine", "learning", "computer"]   # tiny vocabulary, just for illustration


def one_hot(word, vocab):
    """Return a length-|vocab| vector holding 1 at `word`'s index and 0 everywhere else."""
    vec = np.zeros(len(vocab))
    vec[vocab.index(word)] = 1.0
    return vec


def cosine_similarity(a, b):
    """cos(theta) = (a . b) / (|a| * |b|) — the ANGLE between two vectors, independent of their length.

    Why cosine and not plain dot product? Word2Vec vectors are not length-normalised during
    training, so a raw dot product would mix up "similar meaning" with "large vector".
    The epsilon guards against division by zero for an all-zero vector.
    """
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-12))


print(f"One-Hot vectors (vocabulary = {VOCAB}):\n")
for word in VOCAB:
    print(f"  {word:<10} {one_hot(word, VOCAB)}")

print("\nCosine similarity between One-Hot vectors:")
sim_synonyms = cosine_similarity(one_hot("machine", VOCAB), one_hot("computer", VOCAB))
sim_other = cosine_similarity(one_hot("machine", VOCAB), one_hot("learning", VOCAB))
print(f"  machine  vs  computer : {sim_synonyms:+.4f}   <- near-synonyms, but scored as unrelated")
print(f"  machine  vs  learning : {sim_other:+.4f}")
print(f"\nConclusion: both are {sim_synonyms:+.1f}. One-Hot cannot represent similarity at all.")
````

## 2. Data Preprocessing

Gensim performs **no** text cleaning for you. It accepts an iterable of *already tokenised* sentences —
`[["word1", "word2"], ["word3", ...]]` — and learns from whatever you hand it, including capitalised
duplicates, punctuation and filler words. The source notebooks called `doc.split()` and moved straight to
training, which runs fine but leaves three real problems:

| Problem | Example | Fix applied below |
| :--- | :--- | :--- |
| Capitalisation splits one word into two types | `"Machine"` and `"machine"` get **separate** vectors | lowercase everything |
| Punctuation glues onto tokens | `"learning,"` ≠ `"learning"` | strip punctuation with a regex |
| Filler words dominate the counts | `is, a, the, of` appear everywhere and mean little | remove stopwords |

````python
# --- Purpose: define the raw text corpus, exactly as it appears in the course notes. ---
# Why it matters: Word2Vec learns purely from co-occurrence statistics, so the corpus *is* the
# training data — no labels are needed, but the corpus determines every result that follows.
# Sentences 1-10 are the exact sentences from the three source notebooks; the rest widen the domain
# so that similarity, outlier and analogy results are actually interpretable.

RAW_CORPUS = [
    # --- Sentences carried over from the three source notebooks ---
    "machine learning transforms artificial intelligence",
    "machine learning uses neural networks",
    "deep learning is a subset of machine learning",
    "neural networks are used in deep learning",
    "machine learning is powerful",
    "machine learning uses data",
    "deep learning uses neural networks",
    "artificial intelligence uses machine learning",
    "python is used for machine learning",
    "deep learning is part of artificial intelligence",
    # --- Additional sentences so the tiny 10-sentence corpus can learn real structure ---
    "machine learning models learn patterns from data",
    "deep learning models use many layers of neurons",
    "neural networks are layers of connected neurons",
    "artificial intelligence is the simulation of human intelligence",
    "python is a popular programming language for data science",
    "data science uses machine learning and statistics",
    "machine learning is a branch of artificial intelligence",
    "deep learning is a branch of machine learning",
    "convolutional neural networks are used for image recognition",
    "recurrent neural networks are used for language processing",
    "natural language processing is a field of artificial intelligence",
    "machine learning models are trained on large datasets",
    "supervised machine learning uses labelled data",
    "unsupervised machine learning finds patterns in unlabelled data",
    "python libraries simplify machine learning and data science",
    "artificial intelligence powers chatbots and search engines",
    "neural networks learn by adjusting weights during training",
    "gradient descent is used to train neural networks",
    "vectors and matrices are used in machine learning",
    "machine learning is used in healthcare finance and education",
]

print(f"Number of raw sentences in the corpus: {len(RAW_CORPUS)}")
print(f"Total characters: {sum(len(s) for s in RAW_CORPUS)}")
````

````python
# --- Purpose: convert raw sentences into clean, lowercase, stopword-free lists of tokens. ---
# Why it matters: every distinct surface form must collapse to ONE token, otherwise a word is
# split across two vectors and half of its context evidence is wasted.

import re

# Function words that carry almost no topical meaning in this corpus.
STOPWORDS = {
    "is", "are", "was", "were", "a", "an", "the", "and", "or", "of", "for",
    "to", "in", "on", "with", "as", "by", "from", "that", "this", "it", "its",
}


def tokenize(sentence):
    """Lowercase -> strip punctuation -> split on whitespace -> drop stopwords.

    Purpose: turn one raw string into the list-of-tokens format that Gensim requires.
    Why it matters: Word2Vec is a pure count-based learner; whatever noise survives this
    function ends up consuming training pairs and distorting the vectors.
    """
    text = sentence.lower()                                    # 1. case-fold the whole sentence
    text = re.sub(r"[^a-z0-9\s]", " ", text)                  # 2. any punctuation becomes a word break
    tokens = text.split()                                     # 3. split on whitespace
    return [token for token in tokens if token not in STOPWORDS]  # 4. drop stopwords


tokenized_corpus = [tokenize(sentence) for sentence in RAW_CORPUS]

print(f"{'#':<4}{'Tokenised sentence'}")
print("-" * 60)
for i, sentence in enumerate(tokenized_corpus, 1):
    print(f"{i:<4}{sentence}")
````

````python
# --- Purpose: build the vocabulary, count word frequencies, and show what `min_count` would prune. ---
# Why it matters: Gensim silently discards rare words, so knowing the frequency distribution tells
# you whether a word you *expect* to query is even present in the model.

from collections import Counter

# Flatten the corpus into one long stream of tokens, then tally each word.
word_counts = Counter(token for sentence in tokenized_corpus for token in sentence)
vocab_size = len(word_counts)

print(f"Vocabulary size: {vocab_size} unique tokens\n")
print(f"{'Rank':<7}{'Word':<18}{'Count':<8}")
print("-" * 33)
for rank, (word, count) in enumerate(word_counts.most_common(20), 1):
    print(f"{rank:<7}{word:<18}{count:<8}")

# `min_count` demonstration: anything seen fewer than MIN_COUNT times is thrown away.
MIN_COUNT = 2
pruned = sorted(word for word, count in word_counts.items() if count < MIN_COUNT)
print(f"\nWords DROPPED at min_count={MIN_COUNT}: {pruned}")
print(f"Vocabulary after pruning: {vocab_size - len(pruned)} words "
      f"({len(pruned) / vocab_size * 100:.0f}% of the vocabulary removed)")
print("Production models use min_count=5 to discard typos and noise; this practical uses")
print("min_count=1 so that no word from the course notes is lost.")
````

## 3. How Word2Vec Learns (Self-Supervised Learning)

A natural question: *how can a neural network train without human labels?*

Word2Vec invents its own labels using a **sliding context window** (the *distributional hypothesis*:
**words that appear in similar contexts have similar meanings** — "coffee" and "tea" both appear near
"drink", "cup", "hot").

- For the word at index $i$ — the **target** — collect every word within distance $k$ — the **context**.
- **Skip-Gram** (`sg=1`): input = target, labels = context words. *"Given `learning`, predict `machine`."*
- **CBOW** (`sg=0`): input = the averaged context, label = the target. *"Given `machine` and `data`, predict `learning`."*

| Aspect | CBOW (`sg=0`) | Skip-Gram (`sg=1`) |
| :--- | :--- | :--- |
| Learns from | Average of context vectors | Each context word separately |
| Training speed | **Faster** (fewer updates) | Slower (many more pairs) |
| Data hunger | Wants a large corpus | Works well on **small** corpora |
| Rare words | Weak — their signal is averaged away | **Better** for rare words & jargon |
| Semantic quality | Coarser, smoother neighbourhoods | Finer detail, better analogies |

````python
# --- Purpose: generate (target, context) training pairs with a sliding window, by hand. ---
# Why it matters: these pairs ARE the labels. Seeing them makes the whole "unsupervised"
# claim concrete: Word2Vec predicts the neighbourhood, and context-sharing words converge together.

# The course notes trained with a window of 2. We use 3 here: a slightly wider window captures the
# topical association (machine ~ neural ~ networks) more reliably on a corpus this small.
WINDOW = 3


def generate_skipgram_pairs(sentences, window=2):
    """Return [(target, context), ...] for every word and each word within `window` distance.

    Purpose: show exactly the data a Skip-Gram model is trained on.
    Why it matters: the number of pairs produced here is the number of training examples
    per epoch, so it is what `epochs` multiplies.
    """
    pairs = []
    for sentence in sentences:
        for i, target in enumerate(sentence):
            # Clamp the window to the sentence bounds, otherwise the first and last
            # words would produce an IndexError when the window hangs over the edge.
            start = max(0, i - window)
            end = min(len(sentence), i + window + 1)
            for j in range(start, end):
                if i != j:                       # a word is never treated as its own context
                    pairs.append((target, sentence[j]))
    return pairs


training_pairs = generate_skipgram_pairs(tokenized_corpus, window=WINDOW)

print(f"Example sentence: {tokenized_corpus[0]}")
print(f"Window size {WINDOW}  ->  up to {WINDOW} words left and {WINDOW} words right of the target\n")
print(f"{'Target (input)':<16} | {'Context (predicted label)':<28}")
print("-" * 48)
for target, context in training_pairs[:14]:
    print(f"{target:<16} | {context:<28}")
print("-" * 48)
print(f"Total (target, context) pairs in the whole corpus: {len(training_pairs)}")
print(f"With epochs=100 the model makes {len(training_pairs) * 100:,} updates in total.")
````

**One detail Gensim does behind your back:** a *dynamic window*. The effective window is randomly shrunk
(typically 1–5, centre-weighted) on every epoch, so a word sitting 1 position away is counted far more
often than one 5 positions away. That is what makes embeddings capture both tight syntactic relations
and broad topical association. It is not exposed as a parameter — it is always on.

**How the labels are turned into gradients:** instead of a softmax over the whole vocabulary (an
$O(V)$ cost per example), Word2Vec scores every word against the true label plus a handful of randomly
sampled "negative" words, and pushes their scores apart. This is **negative sampling**
(`negative=5` by default in Gensim) and it is the reason training is fast enough to run on billions of
tokens. Both architectures share it.

## 4. Training with Gensim — Hyper-parameter Deep Dive

| Parameter | Value used | Meaning and reasoning |
| :--- | :--- | :--- |
| `sentences` | our tokenised corpus | Iterable of token lists. **Gensim does not clean this for us.** |
| `vector_size` | `50` | Dimensionality of the embedding space. 50 is plenty for a 70-word vocabulary; real models use 100–300. |
| `window` | `3` | Max target↔context distance. A small window (2) captures grammatical/functional relations, 5–10 captures topical association; **3** is the middle ground that keeps `machine` beside `neural` on a corpus this small. |
| `min_count` | `1` | Words seen fewer than this many times are dropped. Must stay `1` here or words leave the vocabulary (§2). |
| `sg` | `0` / `1` | `0` = CBOW, `1` = Skip-Gram. Both are trained so they can be compared. |
| `epochs` | `100` | Full passes over the corpus. The default 5 is far too few for a 30-sentence corpus, so the vectors would still be random noise. |
| `seed` | `42` | Fixes initial weights → reproducible results. |
| `workers` | `1` | Parallel threads. **`workers=1` is required for reproducible training**; higher values are faster but non-deterministic. |
| `negative` | `5` (default) | Number of "negative" words sampled per positive update. |
| `alpha` | `0.025` (default) | Initial learning rate; decays linearly to `min_alpha` over training. |

````python
# --- Purpose: train a CBOW and a Skip-Gram model on the same corpus with identical settings. ---
# Why it matters: holding data and hyper-parameters constant isolates the ARCHITECTURE as the only
# variable, which is the whole point of the comparison in §13.

COMMON_ARGS = dict(
    vector_size=50,   # each word becomes a 50-number dense vector
    window=WINDOW,    # context radius, as in the hand-written pair generation above
    min_count=1,      # keep every word of the practical's vocabulary
    epochs=100,       # many passes: the corpus is tiny, so the default 5 would under-train badly
    seed=42,          # reproducible weight initialisation
    workers=1,        # 1 thread = deterministic; use 4+ for speed on large corpora
)

# sg=0 -> CBOW (predict the target from its context); sg=1 -> Skip-Gram (predict context from target).
cbow_model = Word2Vec(sentences=tokenized_corpus, sg=0, **COMMON_ARGS)
skipgram_model = Word2Vec(sentences=tokenized_corpus, sg=1, **COMMON_ARGS)

model = cbow_model   # the reference model for the rest of the notebook

print("CBOW model (sg=0):")
print(cbow_model)
print("\nSkip-Gram model (sg=1):")
print(skipgram_model)
````

## 5. Inspecting the Learned Vocabulary

During training Gensim builds the vocabulary first: it counts every token, discards those below
`min_count`, sorts the survivors by descending frequency, and assigns each one an integer index.
Those indices are the rows of the embedding matrix.

- `model.wv.index_to_key` — vocabulary words **ordered by frequency** (index `0` is the most common word).
- `model.wv.key_to_index` — the reverse lookup, word → integer index.
- `model.wv.get_vecattr(word, "count")` — how many times the word occurred in the training corpus.
- `model.wv.key_to_index["neural"]` — the row index of a word's vector, useful for slicing the matrix.

````python
# --- Purpose: inspect what the model actually learned — vocabulary size, order and frequencies. ---
# Why it matters: if a word you expect is missing, every later query (`wv[word]`) raises a KeyError,
# so checking the vocabulary is the first debugging step for any Word2Vec script.

vocab = model.wv.index_to_key

print(f"Total vocabulary size: {len(vocab)} words")
print(f"\nVocabulary in descending frequency order:\n{vocab}\n")
print("=" * 46)
print(f"{'Index':<8}{'Word':<20}{'Corpus count':<14}")
print("=" * 46)
for index, word in enumerate(vocab):
    count = model.wv.get_vecattr(word, "count")
    print(f"{index:<8}{word:<20}{count:<14}")

# The integer index of a word is its row inside the embedding matrix.
print("\n" + "=" * 46)
for word in ["machine", "learning", "neural"]:
    if word in model.wv:
        print(f"'{word}' -> index {model.wv.key_to_index[word]}")
    else:
        print(f"'{word}' -> NOT in vocabulary (dropped by min_count)")
````

## 6. Extracting Word Embeddings (Dense Vectors)

`model.wv[word]` returns one row of the embedding matrix as a **NumPy array of 50 float32 values**.

- Every coordinate is a *latent* feature discovered by gradient descent — it is not a human-readable
  property like "is_animal". Interpreting individual dimensions is not the point; the geometry of the
  whole vector is.
- Because the values are unnormalised, `most_similar` (and Gensim's `similarity`) applies cosine
  similarity, which ignores vector length and looks only at direction.
- **De-duplication note:** the source notebooks extracted `wv["machine"]` in two identical consecutive
  cells. That duplication is removed — there is one extraction and one analysis cell below.

````python
# --- Purpose: extract one word vector and report its shape, dtype, norm and statistics. ---
# Why it matters: understanding the numbers confirms the model really produced dense, continuous
# values (as opposed to the sparse 0/1 vectors of Practicals 1-3).

target_word = "machine"
vector = model.wv[target_word]     # one row of the embedding matrix, shape (50,)

print(f"Vector for '{target_word}':")
print(f"  shape   : {vector.shape}   -> 50 dense dimensions")
print(f"  dtype   : {vector.dtype}    -> continuous floating point, not sparse 0/1")
print(f"  L2 norm : {np.linalg.norm(vector):.4f}  (vector length)")
print(f"  values  : {np.array2string(vector[:10], precision=4)} ... (first 10 of 50)")

print("\nSummary statistics of the embedding weights:")
print(f"  min  : {vector.min():+.4f}")
print(f"  max  : {vector.max():+.4f}")
print(f"  mean : {vector.mean():+.4f}")
print(f"  std  : {vector.std():.4f}")
print(f"  zeros: {int((vector == 0).sum())} of {vector.size}  -> 0% sparse")
````

## 7. Semantic Similarity & Nearest Neighbours

Two words are semantically related when their vectors point in a similar **direction**:

$$\text{cos}(\mathbf{u}, \mathbf{v}) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\|_2\,\|\mathbf{v}\|_2}$$

| Score | Interpretation |
| :--- | :--- |
| `+1.0` | Identical direction — maximum similarity |
| `0.0` | Orthogonal — unrelated |
| `-1.0` | Opposite directions |

- `model.wv.most_similar(word, topn=N)` — cosine similarity of `word` against **every** other word, top `N` returned.
- `model.wv.similarity(w1, w2)` — the single cosine score for a pair.

````python
# --- Purpose: find the nearest semantic neighbours and score specific word pairs. ---
# Why it matters: this is the payoff of Word2Vec — words that share contexts end up close together,
# and the manual NumPy implementation is cross-checked against Gensim's own answer to prove it.

# 1. Nearest neighbours of 'machine'
print("Top 5 words most similar to 'machine':")
for rank, (word, score) in enumerate(model.wv.most_similar("machine", topn=5), 1):
    print(f"  {rank}. {word:<20} cosine = {score:+.4f}")

# 2. Pairwise scores between domain terms
print("\nPairwise cosine similarities:")
pairs = [
    ("machine", "learning"),
    ("deep", "learning"),
    ("artificial", "intelligence"),
    ("neural", "networks"),
    ("python", "machine"),
    ("data", "python"),
]
for w1, w2 in pairs:
    print(f"  {w1:<12} vs {w2:<14} cosine = {model.wv.similarity(w1, w2):+.4f}")

# 3. Verify Gensim against the NumPy formula from §1 — if these two agree, our
#    understanding of cosine similarity is correct (allclose tolerates float rounding).
def cosine_similarity(a, b):
    """Reused from §1: the angle between two vectors, independent of their length."""
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-12))


manual = [cosine_similarity(model.wv[w1], model.wv[w2]) for w1, w2 in pairs]
gensim = [model.wv.similarity(w1, w2) for w1, w2 in pairs]
print(f"\nNumPy implementation matches Gensim: {np.allclose(manual, gensim, atol=1e-5)}")
````

## 8. Semantic Outlier Detection with `doesnt_match`

`model.wv.doesnt_match(words)` returns the word that belongs least in a list. The maths is simple:

1. Look up every **in-vocabulary** word: $\mathbf{v}_1, \dots, \mathbf{v}_n$.
2. Compute the centroid $\mathbf{c} = \frac{1}{n}\sum_i \mathbf{v}_i$.
3. Score each word by $\text{cos}(\mathbf{v}_i, \mathbf{c})$.
4. Return the word with the **lowest** score.

**What happened in the source notebooks:** `doesnt_match(["machine", "learning", "deep", "apple"])`
returned `'learning'`, not `'apple'` — because `apple` was never in the training corpus. Gensim logs a
warning and **silently drops** out-of-vocabulary inputs, so only three words were actually compared.
That is a real gotcha: an OOV word can never be the answer, and the "outlier" you get back is
arbitrary. §10 shows how to check membership yourself.

````python
# --- Purpose: demonstrate outlier detection, and expose how out-of-vocabulary words are ignored. ---
# Why it matters: `doesnt_match` is only as trustworthy as the vocabulary behind it, so we verify
# membership before trusting the answer.

# Test 1: a group of genuine in-vocabulary terms
group_1 = ["machine", "learning", "neural", "data"]
print(f"Input: {group_1}")
print(f"  all in vocabulary : {all(w in model.wv for w in group_1)}")
print(f"  outlier           : '{model.wv.doesnt_match(group_1)}'\n")

# Test 2: a second grouping
group_2 = ["python", "data", "statistics", "neural"]
print(f"Input: {group_2}")
print(f"  all in vocabulary : {all(w in model.wv for w in group_2)}")
print(f"  outlier           : '{model.wv.doesnt_match(group_2)}'\n")

# Test 3: the OOV case from the source notebooks
group_3 = ["machine", "learning", "deep", "apple"]
print(f"Input: {group_3}")
print(f"  'apple' in vocabulary: {'apple' in model.wv}")
print(f"  outlier returned     : '{model.wv.doesnt_match(group_3)}'")
print("  -> 'apple' was discarded as out-of-vocabulary, so the answer comes from the")
print("     remaining words only. This is NOT a reliable outlier result.")
````

## 9. Vector Arithmetic & Word Analogies

The celebrated property of embeddings is that **linear operations approximate semantic relations**:

$$\vec{v}_{\text{king}} - \vec{v}_{\text{man}} + \vec{v}_{\text{woman}} \approx \vec{v}_{\text{queen}}$$

Subtracting `man` is read as removing a "male" direction; adding `woman` adds a "female" direction.
Gensim exposes this through `most_similar(positive=[...], negative=[...])` — add the positives,
subtract the negatives, then return the nearest words to the result.

**Honest limitation:** the classic `king : man :: woman : queen` test cannot be run here, because
`king`, `man`, `woman` and `queen` do not occur in this corpus. Analogy quality needs millions of
tokens. The two queries below therefore use domain terms that *are* in the vocabulary:
`data + science` (pure addition) and `neural + networks − machine` (addition **and** subtraction).
The second is the interesting one — removing the "machine" direction moves the result out of the
machine-learning cluster and towards the artificial-intelligence cluster.

````python
# --- Purpose: perform semantic vector arithmetic and interpret the resulting neighbours. ---
# Why it matters: it demonstrates the one capability no sparse representation (BoW/TF-IDF) can offer,
# while keeping expectations honest about what a 30-sentence corpus can support.


def show_query(positive, negative, model, label):
    """Print the top neighbours of (sum of positive vectors) - (sum of negative vectors)."""
    print(label)
    for rank, (word, score) in enumerate(
        model.wv.most_similar(positive=positive, negative=negative, topn=4), 1
    ):
        print(f"  {rank}. {word:<16} cosine = {score:+.4f}")
    print()


# 1. Addition only: what sits next to the "data science" concept?
show_query(["data", "science"], [], model, "(+data +science)  --  pure addition")

# 2. Addition and subtraction: strip the "machine" direction out of "neural networks".
#    Expected reading: the result drifts out of the machine-learning cluster and towards
#    the artificial-intelligence cluster (chatbots, powers, intelligence).
show_query(["neural", "networks"], ["machine"], model, "(+neural +networks) -machine  --  addition + subtraction")

# 3. The same two queries on the Skip-Gram model, to compare the architectures.
show_query(["data", "science"], [], skipgram_model, "Skip-Gram: (+data +science)")
show_query(["neural", "networks"], ["machine"], skipgram_model, "Skip-Gram: (+neural +networks) -machine")
````

## 10. Handling Out-of-Vocabulary (OOV) Words Safely

Word2Vec works at the **word level** against a vocabulary frozen at training time. Querying a word that
never appeared in the corpus — `"transformer"`, `"chatgpt"`, a typo like `"learnng"` — raises a hard
`KeyError` and stops the cell.

Always guard your lookups:

```python
if word in model.wv:            # membership test
    vector = model.wv[word]
else:                            # or catch the error
    vector = None
```

This limitation is what later architectures fixed: **FastText** (2017) represents a word as a bag of
character n-grams, so unseen words still get sensible vectors, and **subword tokenisation**
(BPE / WordPiece) made it standard in Transformers such as BERT and GPT.

````python
# --- Purpose: show the KeyError that an unguarded lookup raises, then handle it safely. ---
# Why it matters: in any real pipeline the user supplies the query words, so an unguarded
# lookup is a crash waiting to happen.

query_words = ["machine", "neural", "transformer", "learnng", "apple"]

print(f"{'Query':<14}{'In vocabulary?':<18}{'Action'}")
print("-" * 60)
for word in query_words:
    if word in model.wv:
        print(f"{word:<14}{'yes':<18}vector retrieved, L2 norm = {np.linalg.norm(model.wv[word]):.3f}")
    else:
        print(f"{word:<14}{'no':<18}skipped (would raise KeyError)")

# Demonstrate the failure mode explicitly, then catch it.
print("\nUnguarded lookup:")
try:
    _ = model.wv["transformer"]
except KeyError as error:
    print(f"  KeyError raised as expected: {error}")

# The safe pattern: a helper that returns None instead of crashing.
def safe_vector(word, wv):
    """Return the vector for `word`, or None when the word was never seen in training."""
    return wv[word] if word in wv else None


print("\nSafe pattern:")
for word in ["machine", "transformer"]:
    vec = safe_vector(word, model.wv)
    print(f"  {word:<14}{'vector found' if vec is not None else 'None returned - no crash'}")
````

## 11. Visualising the Embedding Space with PCA

The vectors live in 50-dimensional space, which no screen can show. **Principal Component Analysis**
finds the two orthogonal directions of maximum variance and projects every vector onto them:

- `pca.fit_transform(...)` — fit on the matrix of all word vectors, return the 2-D coordinates.
- `pca.explained_variance_ratio_` — how much of the original variance each axis captures.

**Read the plot with care:** a 2-D projection necessarily distorts distances. Two words that look close
here may not be close in the full 50-D space, and vice versa. The plot is for *qualitative* insight
(are the ML terms grouped together?), not for measuring similarity — use `most_similar` for that.

````python
# --- Purpose: project all word vectors into 2-D with PCA and plot them as a labelled scatter. ---
# Why it matters: it makes the central claim of Word2Vec visible — words used in similar contexts
# occupy the same region of vector space.

words = list(model.wv.index_to_key)
word_vectors = np.array([model.wv[w] for w in words])     # shape (n_words, 50)

# Fit PCA on the (n_words x 50) matrix and project down to 2 dimensions.
pca = PCA(n_components=2, random_state=42)
coords_2d = pca.fit_transform(word_vectors)

pc1_var = pca.explained_variance_ratio_[0] * 100
pc2_var = pca.explained_variance_ratio_[1] * 100

plt.figure(figsize=(11, 7))
plt.scatter(coords_2d[:, 0], coords_2d[:, 1], s=130, color="#2b5c8f", edgecolors="black", alpha=0.85)

# Annotate each point with its word so the clusters are readable.
for i, word in enumerate(words):
    plt.annotate(word, xy=(coords_2d[i, 0], coords_2d[i, 1]), xytext=(7, 4),
                 textcoords="offset points", fontsize=10)

plt.title("2-D PCA projection of the Word2Vec embedding space", fontsize=14, fontweight="bold", pad=15)
plt.xlabel(f"Principal Component 1 ({pc1_var:.1f}% of total variance)", fontsize=11)
plt.ylabel(f"Principal Component 2 ({pc2_var:.1f}% of total variance)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print(f"Total variance retained by the 2-D projection: {pc1_var + pc2_var:.2f}%")
print("A low percentage is expected: most of the information in 50 dimensions cannot fit in 2.")
````

## 12. Saving, Loading & Reproducibility

Training a real Word2Vec model takes hours, so it is trained once and then serialised. Two levels exist:

| Call | Saves | Use when |
| :--- | :--- | :--- |
| `model.save("m.model")` | Full model: vocabulary, window, `seed`, and all weights | You may want to **continue training** (`.build_vocab()` / `.train()`) |
| `model.wv.save("m.kv")` | Only the `KeyedVectors` (the vectors + vocabulary) | You only need lookup / similarity — a much smaller file |

**Reproducibility checklist**

- `seed=42` fixes the initial weights.
- `workers=1` is mandatory — multi-threaded training updates weights in a non-deterministic order.
- For results identical across *separate sessions*, also start Jupyter with `PYTHONHASHSEED=0`,
  because Python's string hashing decides the vocabulary ordering.

````python
# --- Purpose: save the model, reload it, prove the reload is lossless, and check determinism. ---
# Why it matters: a model you cannot reload is not a deliverable, and results you cannot
# reproduce cannot be marked correct in a lab submission.

import os

MODEL_FILE = "word2vec_model.model"

# 1. Serialise the trained model to disk.
model.save(MODEL_FILE)
print(f"Saved '{MODEL_FILE}' ({os.path.getsize(MODEL_FILE) / 1024:.2f} KB)")

# 2. Reload it in a fresh object.
loaded_model = Word2Vec.load(MODEL_FILE)
print("Reloaded successfully from disk.")

# 3. Lossless round trip: every reloaded vector must equal the original.
all_match = all(
    np.allclose(loaded_model.wv[word], model.wv[word], atol=1e-6)
    for word in model.wv.index_to_key
)
print(f"All {len(model.wv.index_to_key)} vectors identical after reload: {all_match}")

# 4. Determinism check: retrain from scratch with the same seed and compare.
retrained = Word2Vec(sentences=tokenized_corpus, sg=0, **COMMON_ARGS)
deterministic = np.allclose(retrained.wv["machine"], model.wv["machine"], atol=1e-6)
print(f"Retraining with seed=42 and workers=1 reproduces the same vector: {deterministic}")

# 5. A sample of the reloaded embeddings.
print("\nSample vectors from the reloaded model:")
for index, word in enumerate(loaded_model.wv.index_to_key[:6], 1):
    sample = np.round(loaded_model.wv[word][:5], 4)
    print(f"  {index}. {word:<14} shape={loaded_model.wv[word].shape}  first 5 dims={sample}")
````

## 13. CBOW vs Skip-Gram, and Honest Limitations

Both models were trained on identical data with identical hyper-parameters, so any difference in the
table below comes from the architecture alone.

````python
# --- Purpose: compare what each architecture learned on the same word pairs. ---
# Why it matters: it shows CBOW averages its context (smoother, higher baseline similarities)
# while Skip-Gram keeps per-pair detail (typically a wider spread of scores).

comparison_pairs = [
    ("machine", "learning"),
    ("deep", "learning"),
    ("artificial", "intelligence"),
    ("neural", "networks"),
    ("python", "machine"),
    ("data", "neural"),
]

print(f"{'Word pair':<34}{'CBOW (sg=0)':<16}{'Skip-Gram (sg=1)':<18}")
print("=" * 68)
for w1, w2 in comparison_pairs:
    sim_cbow = cbow_model.wv.similarity(w1, w2)
    sim_sg = skipgram_model.wv.similarity(w1, w2)
    print(f"{f'({w1!r}, {w2!r})':<34}{sim_cbow:<+16.4f}{sim_sg:<+18.4f}")

print(f"\nMean pairwise similarity  CBOW: "
      f"{np.mean([cbow_model.wv.similarity(a, b) for a, b in comparison_pairs]):+.4f}")
print(f"Mean pairwise similarity  Skip-Gram: "
      f"{np.mean([skipgram_model.wv.similarity(a, b) for a, b in comparison_pairs]):+.4f}")
````

### Limitations of what we just built

1. **The corpus is far too small.** ~30 sentences and ~70 words. Similarity values are unstable and
   the analogies are suggestive, not reliable. Real Word2Vec models train on billions of tokens.
2. **Static vectors — no polysemy.** The word *bank* gets exactly one vector, whether it means
   financial bank or river bank. Contextual models (ELMo, BERT) fix this.
3. **Out-of-vocabulary words are fatal.** `wv[word]` raises `KeyError`; FastText and subword models solve it.
4. **Rare words get poor vectors.** Their few co-occurrences are not enough to place them well.
5. **Similarity is not entailment.** High cosine similarity means similar *usage*, not logical implication.

| Property | One-Hot | BoW | TF-IDF | **Word2Vec** |
| :--- | :--- | :--- | :--- | :--- |
| Vector type | Sparse, binary | Sparse counts | Sparse weights | **Dense, continuous** |
| Size | $\|V\|$ | $\|V\|$ | $\|V\|$ | **50–300 (fixed)** |
| Captures similarity | No (`0.0`) | No (`0.0`) | Statistical only | **Yes** |
| Context aware | No | No | No | **Yes (sliding window)** |
| Word order | No | No | No | **Yes (local)** |
| Vector arithmetic | No | No | No | **Yes** |
| OOV words | Crash | Ignored | Ignored | **Needs a check (§10)** |

## 14. Summary & Key Takeaways

1. **Preprocessing is your job.** Gensim consumes pre-tokenised text; case, punctuation and stopwords
   all reach the vectors unless you remove them (§2).
2. **Word2Vec is self-supervised.** The sliding window manufactures the labels: every `(target, context)`
   pair is a training example, and context-sharing words converge (§3).
3. **CBOW vs Skip-Gram** is a speed/data trade-off: CBOW averages context and is faster; Skip-Gram
   trains on individual pairs and handles rare words better (§4, §13).
4. **Similarity is direction, not length** — cosine similarity, verified against Gensim's own
   implementation in §7.
5. **`workers=1` + `seed` + `PYTHONHASHSEED=0` is the recipe for reproducible embeddings** (§12).
6. **Know the failure modes**: tiny corpora, OOV lookups, polysemy (§13).

§4 train both models → §5 vocabulary → §6 extract vectors → §7 similarity → §8 outliers →
§9 analogies → §10 OOV safety → §11 PCA plot → §12 save/load → §13 comparison & limits.
