```python
# import TfidfVectorizer from sklearn
from sklearn.feature_extraction.text import TfidfVectorizer

# Pandas for displaying the TF-IDF matrix
import pandas as pd
print("Libraries Imported Successfully!")
```

```python
# Step 1: Create the Corpus

corpus = ["The quick brown fox jumps over the lazy dog.",
    "The dog was not lazy, the dog was just tired.",
    "A quick fox can jump higher than a lazy dog.",
    "The brown fox and the blue bird are friends.",
    "Can a dog jump over a fox? Yes, a dog can.",
    "The tired bird rests on the brown branch.",
    "Lazy dogs sleep all day, but quick foxes run.",
         ]
print("Corpus:\n", corpus)
```

```python
for i, sentence in enumerate(corpus):
    print(f"Document: {i+1}: {sentence}")
```

```python
# Step 2: Total Number of Documents
total_documents = len(corpus)

print("Total Documents =", total_documents)
```

```python
# Step 3: Create a TF-IDF object
tfidf = TfidfVectorizer()
print(tfidf)
```

```python
# Step 4: Learn Vocabulary

# Learn vocabulary and calculate TF-IDF values
tfidf_matrix = tfidf.fit_transform(corpus)
print("TF IDF learned Successfully!")
```

```python
# Step 5: Display Vocabulary

vocabulary = tfidf.get_feature_names_out()
print("Vocabulary:\n")

for word in vocabulary:
    print(word)
```

```python
print("Shape of TF-IDF Matrix")
print(tfidf_matrix.shape)
```

```python
dense_matrix = tfidf_matrix.toarray()
print(dense_matrix)
```

```python
# Displaqy TF-IDF Matix as Dataframe
df1=pd.DataFrame(dense_matrix,columns=vocabulary)
```

```python
df = pd.DataFrame(
    dense_matrix,
    columns=vocabulary)

df.index = [
    "Document 1",
    "Document 2",
    "Document 3",
    "Document 4",
    "Document 5",
    "Document 6",
    "Document 7",
]

print(df)
df
```
